```python
import requests
import os
```

In ./app folder:
`uvicorn main:app --reload`

```python
# /upload test
video_to_upload_path = '.temp/IMG_h1.MOV' 

if not os.path.exists(video_to_upload_path):
    print(f"Error: Video file not found at {video_to_upload_path}. Please update the path.")
else:
    upload_url = 'http://127.0.0.1:8000/api/upload'
    with open(video_to_upload_path, 'rb') as f:
        upload_files = {'video': (os.path.basename(video_to_upload_path), f, 'video/mp4')}
        upload_response = requests.post(upload_url, files=upload_files)

    print(f'Upload Status Code: {upload_response.status_code}')
    print(f'Upload Response JSON: {upload_response.json()}')
```

```python
# /process/{process_id} test
if upload_response.status_code == 200:
    upload_data = upload_response.json()
    process_id = upload_data.get('pid')

    if process_id:
        process_url = f'http://127.0.0.1:8000/api/process/{process_id}'
        process_payload = {'exercise_name': 'handstand'}
        
        process_response = requests.post(process_url, json=process_payload)

        print(f'Process Status Code: {process_response.status_code}')
        print(f'Process Response JSON: {process_response.json()}')
    else:
        print("Error: No process_id received from upload.")
else:
    print("Error: Video upload failed.")
```

### Status and Download Tests
Run the following cells after running the `/upload` and `/process` tests above.

```python
# GET /status/{process_id} test
# You can run this cell multiple times to check the processing status.
if 'process_id' in locals() and process_id:
    status_url = f'http://127.0.0.1:8000/api/status/{process_id}'
    status_response = requests.get(status_url)
    
    print(f'Status Response Code: {status_response.status_code}')
    print(f'Status Response JSON: {status_response.json()}')
else:
    print("Please run the '/upload' and '/process' cells first to get a process_id.")
```

```python
# GET /download/{process_id}/json test
if 'process_id' in locals() and process_id:
    download_json_url = f'http://127.0.0.1:8000/api/download/json/{process_id}'
    download_json_response = requests.get(download_json_url)

    if download_json_response.status_code == 200:
        print("Successfully downloaded JSON output:")
        print(download_json_response.json())
    else:
        print(f"Error downloading JSON: {download_json_response.status_code}")
        print(download_json_response.text)
else:
    print("Please run the '/upload' and '/process' cells first to get a process_id.")
```

```python
# GET /download/{process_id}/video test
if 'process_id' in locals() and process_id:
    download_video_url = f'http://127.0.0.1:8000/api/download/video/{process_id}'
    download_video_response = requests.get(download_video_url)

    if download_video_response.status_code == 200:
        # Save the video
        saved_video_path = 'downloaded_video.mp4'
        with open(saved_video_path, 'wb') as f:
            f.write(download_video_response.content)
        print(f"Successfully downloaded and saved video to {os.path.abspath(saved_video_path)}")
    else:
        print(f"Error downloading video: {download_video_response.status_code}")
        print(download_video_response.text)
else:
    print("Please run the '/upload' and '/process' cells first to get a process_id.")
```

```python
# GET /download/{process_id}/pdf test
if 'process_id' in locals() and process_id:
    download_pdf_url = f'http://127.0.0.1:8000/api/download/pdf/{process_id}'
    download_pdf_response = requests.get(download_pdf_url)

    if download_pdf_response.status_code == 200:
        # Save the PDF
        saved_pdf_path = 'downloaded_report.pdf'
        with open(saved_pdf_path, 'wb') as f:
            f.write(download_pdf_response.content)
        print(f"Successfully downloaded and saved PDF to {os.path.abspath(saved_pdf_path)}")
    else:
        print(f"Error downloading PDF: {download_pdf_response.status_code}")
        print(download_pdf_response.text)
else:
    print("Please run the '/upload' and '/process' cells first to get a process_id.")
```

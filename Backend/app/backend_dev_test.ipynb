```python
import json

from repositories.MediapipeSegmentationRepository import MediapipeSegmentationRepository
from services.AnalyticalModel import AnalyticalModel
from services.VisualisationService import VisualisationService
```

```python
input_video_path = "../backend_dev_test_data/test_input.mov"
output_video_path = "../backend_dev_test_data/test_output.mp4"
```

```python
segmentor = MediapipeSegmentationRepository()
analytical_model = AnalyticalModel(reference_json_path="../backend_dev_test_data/test_reference.json")
```

```python
reference_videos_paths = [f"../backend_dev_test_data/poses_references/{i}" for i in range(3)]
    
swing_arms_config = {
    "is_sub_pose": True,
    "segment_mask": [0, 1, 2, 3],
    "feature_mask": [False, True, True]
}

swing_back_config = {
    "is_sub_pose": True,
    "segment_mask": [1, 7, 2, 8],
    "feature_mask": [False, True, True]
}

swing_composite_config = {
    "sub_poses": ["swing_back", "swing_arms"]
}

analytical_model.fit("starting", reference_videos_paths[0])

analytical_model.fit("swing_arms", reference_videos_paths[1], swing_arms_config)
analytical_model.fit("swing_back", reference_videos_paths[1], swing_back_config)
analytical_model.fit("swing", None, swing_composite_config)

analytical_model.fit("landing", reference_videos_paths[2])
```

```python
frames, tensor_3d, mask = segmentor.process_video(input_video_path)
poses = analytical_model.get_poses_from_video(input_video_path)
```

```python
for pose in poses:
   print(json.dumps(pose, indent=2))
   print("-" * 8) 
```

```python
poses_formated = []
for pose in poses:
    pose = dict(sorted(pose.items(), key=lambda item: item[1]["score"], reverse=True))
    pose_name, pose_data = list(pose.items())[0]
    result = f"{pose_name}: {pose_data['score']:.2f}"
    
    for sub_pose_name, sub_pose_score in pose_data["sub_scores"].items():
        result += f" | {sub_pose_name}: {sub_pose_score:.2f}"
    
    poses_formated.append(result)
```

```python
print(poses_formated)
```

```python
VisualisationService.save_visualized_video(output_video_path, frames, pose_labels=poses_formated, input_video_path=input_video_path)
```
